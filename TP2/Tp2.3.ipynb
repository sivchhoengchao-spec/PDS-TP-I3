```python
#step1
student = ("Dara", 20, 3.45) 
print("Name:", student[0]) 
```

```python
#step2
name, age, gpa = student 
print(name, "is", age, "years old, GPA", gpa)
```

```python
#step3
students = [("Dara", 20, 3.45), ("Sophea", 21, 3.92), 
            ("Vibol", 19, 3.10), ("Malis", 22, 3.78)] 
for name, age, gpa in students: 
    print(name, age, gpa) 
```

```python
#step4
ranking = sorted(students, key=lambda s: s[2], reverse=True)
print("Best student:", ranking[0][0]) 
```
