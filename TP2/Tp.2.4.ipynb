```python
#step1
football_list = ["Dara", "Vibol", "Kosal", "Dara", "Bopha"] 
football = set(football_list) 
print("List size:", len(football_list), "| Set size:", len(football))
```

```python
#step2
music = {"Sophea", "Kosal", "Malis", "Dara"}
```

```python
#step3
print("Both clubs    :", sorted(football & music))  
print("Any club      :", sorted(football | music))   
print("Football only :", sorted(football - music))  
print("Exactly one   :", sorted(football ^ music))  
```

```python
#step4
music.add("Chenda") 
music.discard("Malis") 
print("Music now     :", sorted(music)) 
print("Is Vibol in music?", "Vibol" in music) 
```

1.  Because a set does not allow duplicate values. "Dara" appears twice in the original list, but the set stores it only once.

2.  The order of a set is not guaranteed because sets are unordered collections.
We cannot use football[0] because sets do not support indexing.

3.  Football club:
- Dara
- Vibol
- Kosal
- Bopha
    Music club:
- Sophea
- Kosal
- Malis
- Dara


```python
import matplotlib.pyplot as plt
from matplotlib_venn import venn2
v = venn2([football, music], set_labels=("Football Club", "Music Club"))

v.get_label_by_id('10').set_text("Vibol\nBopha")
v.get_label_by_id('11').set_text("Dara\n Kosal")
v.get_label_by_id('01').set_text("Sophea\n Malis")
                  
plt.show()
```
