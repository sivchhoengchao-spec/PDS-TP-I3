```python
from pathlib import Path
import pandas as pd
import numpy as np

DATA = Path.cwd()

reg = pd.read_csv(DATA / "registrations.csv") 

print("Duplicated rows:", reg.isna().duplicated().sum()) 
```

```python
#step2

reg["name"] = reg["name"].str.strip().str.lower() 
reg["city"] = reg["city"].str.strip().str.title() 
print("Duplicated rows:", reg.duplicated().sum())
```

```python
#step3

reg = reg.drop_duplicates().reset_index(drop=True) 
print(reg) 
```

```python
#step5

print("Sum as text  :", reg["fee"].sum()) 
reg["fee"] = reg["fee"].str.replace("$", "").astype(float) 
print("Sum as number:", reg["fee"].sum())
```
