```python
#step1

from pathlib import Path
import pandas as pd
import numpy as np

DATA = Path.cwd()

df = pd.read_csv(DATA / "students.csv") 

print(df.groupby("major")["gpa"].mean().round(2))
```

```python
#step 2
#count non missing valuse
summary = df.groupby("major").agg( 
    students=("name", "count"), 
    mean_gpa=("gpa", "mean"), 
    max_gpa=("gpa", "max"), 
    mean_age=("age", "mean"), 
).round(2) 
print(summary) 
```

```python
#step 3

print(df.groupby(["major", "gender"])["gpa"].mean().round(2).unstack()) 

```

```python
#step4

best_rows = df.groupby("major")["gpa"].idxmax() 
print(df.loc[best_rows, ["major", "name", "gpa"]])


```

```python
#count all rows
summary = df.groupby("major").agg( 
    students=("name", "size"), 
    mean_gpa=("gpa", "mean"), 
    max_gpa=("gpa", "max"), 
    mean_age=("age", "mean"), 
).round(2) 
print(summary) 
```
