```python
#step1
from pathlib import Path
import pandas as pd

DATA = Path.cwd()

df = pd.read_csv(DATA / "students.csv")

print(df[df["gpa"]>3.7][["name","gpa"]])
```

```python
#step2

cs_young = df[(df["major"]=="CS") & (df["age"]<22)]
print(cs_young[["name", "major", "age"]])
```

```python
#step3

south = df[df["city"].isin(["Siem Reap","Kampot"])]
print("Siem Reap or Kompot: ", south["name"].tolist())
```

```python
#step4

print(df.sort_values("gpa", ascending = False ).head(3)[["name","gpa"]])
```

```python
#step5

ranked = df.sort_values(["major", "gpa"], ascending = [True,False])
print(ranked["name"].tolist()) 
```
