```python
#step1
from pathlib import Path
import pandas as pd

DATA = Path.cwd()

sales = pd.read_csv(DATA / "cafe_sales.csv", index_col = "day")["cups"]

print(sales.head(3))
```

```python
#step2

print("Total : ",sales.sum()) 
print("Mean  :", round(sales.mean(), 1)) 
print("Median:", sales.median()) 
```

```python
#step3

busy = sales[sales >= 150] 
print("Busy days :", busy.index.tolist()) 
print("Quiet days:", (sales <150).sum()) 
```

```python
#step4

top3 = sales.sort_values(ascending=False).head(3) 
print("Top 3:", top3.to_dict())
```

```python
#step5

weekend = sales[["Sat", "Sun"]].sum()
print("Weekend share:", round(weekend / sales.sum() * 100, 1), "%") 
```
