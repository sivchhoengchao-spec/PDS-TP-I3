```python
#step1

from pathlib import Path
import pandas as pd
import numpy as np

DATA = Path.cwd()

df = pd.read_csv(DATA / "students.csv") 
df["gpa_percent"] = (df["gpa"] / 4.0* 100).round(1) 
```

```python
#step2

df["grade"] =np.where(df["gpa"]>=3.7,"A",np.where(df["gpa"]>=3.3,"B","C"))
```

```python
#step3

def age_group(age):
    if age <20:
        return "teen"
    elif age <=21:
        return "20-21"
    else:
        return "22+"

df["age_group"] = df["age"].apply(age_group)
```

```python
#step4

df["code"] = df["major"] + "-" + df["id"].astype(str)
```

```python
#step5

print(df[["name", "gpa_percent", "grade", "age_group", "code"]])
```
