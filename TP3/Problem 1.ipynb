```python
#step1
from pathlib import Path
import pandas as pd

DATA = Path.cwd()
prices = pd.read_csv( DATA /  "prices.csv", index_col = "product")["price"]
print(prices)
```

```python
#step2

print("price of a pen :", prices["pen"])
print("Number of product : ", len(prices))
```

```python
#step3

print("Most expensive:", prices.idxmax(), prices.max()) 
print("Cheapest      :", prices.idxmin(), prices.min())
```

```python
#step4

mean = prices.mean() 
print("Mean price:", round(mean, 2)) 
print("Above the mean:", prices[prices>mean].index.tolist())
```

```python
#step5

print("Pen in riel:", prices["pen"] * 4100) 
print("After 10% off:", (prices * 0.9).round(2).tolist())
```
