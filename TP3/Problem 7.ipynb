```python
from pathlib import Path
import pandas as pd
import numpy as np

DATA = Path.cwd()

survey = pd.read_csv(DATA / "survey.csv") 

print("Missing per column:", survey.isna().sum().to_dict()) 
```

```python
#step2

holes = survey[survey.isna().any(axis=1)] 
print("Rows with holes:", holes["name"].tolist()) 
print("dropna() keeps", len(survey.dropna()), "of", len(survey), "rows")
```

```python
#step3

survey["age"] = survey["age"].fillna(survey["age"].median()) 
```

```python
#step4

survey["gpa"] = survey["gpa"].fillna(round(survey["gpa"].mean(), 2)) 
survey["city"] = survey["city"].fillna(survey["city"].mode()[0]) 
```

```python
#step5

print(survey) 
print("Missing values left:", survey.isna().sum().sum())
```
