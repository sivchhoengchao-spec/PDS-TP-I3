```python
print("Problem 3. Temperature Converter")
def celsius_to_fahrenheit (c,round_to=1) :
    F= c*9/5 +32
    return round(F,round_to)
def fahrenheit_to_celsius(f, round_to=1):
    C = (f-32)*5/9
    return round(C, round_to)
print("°C      |     °F")
print("--------+-------")
for c in range(0,101,20):
     print(c,"\t|    ",celsius_to_fahrenheit(c))

print("--------+-------")
print(celsius_to_fahrenheit(37),"°F = ",fahrenheit_to_celsius(98.6),"°C ")


print ("37°C (round_to=3) = ",celsius_to_fahrenheit(37, round_to=3),"°F")


print("A default argument is useful because we don't need to specify round_to every time.\nIt automatically uses 1 decimal place.")
```

```python
print("Problem 6. Leap Years and Days in a Month")
def leap_year(year):
    if (year%400==0 )or(year %4==0 and year%100):
        return True
    else:
        return False
def day_in_month(month,year):
    if month in (4,6,9,11):
        return 30
    elif month ==2:
        if leap_year(year)==True:
            return 29
        else :
            return 28
    elif  month in (1,3,5,7,8,10,12):
        return 31
    else:
        return None
    
years = [1900, 2000,2024,2026]
for year in years:
    print(year ,":", "leap" if leap_year(year)==True else "NOt leap")
print("\nFeb 2024 :" ,day_in_month(2,2024))
print("Feb 2026 :" ,day_in_month(2,2026))
print("Apr :" ,day_in_month(4,2024))
print("month 13 :" ,day_in_month(13,2024))
count=0
for year in range(2000,2100):
    if leap_year(year)==True:
        count+=1
print("Leap year form 2000 to 2100 :", count)



```

```python
print("Problem 7. Digit Tools ")
def digit_sum(n):
    n = abs(n)
    sum =0
    while n>0:
        sum = sum + (n%10)
        n =n//10
    return sum


def reverse_number(n):
    reverses =0
    while n>0:
        digit= n%10
        reverses =reverses *10 + digit
        n=n//10
    return reverses


def is_palindrome(n):
    return n==reverse_number(n)


for i in (2026,12321,9875):
    print(i ,":","  digit sum =",digit_sum(i),",reversed =",reverse_number(i),",palidrome =",is_palindrome(i))
    count=0
for  i in range(10,1000):
    if is_palindrome(i):
        count+=1
print("palidromes from 10 to 999 :",count)
```

```python
from numpy import mean, number
print("Problem 8. Flexible Functions: *args, **kwargs and lambda")
# ---------------------------------------------------------------------------
# Problem 8 (Medium) · *args, **kwargs and lambda
# ---------------------------------------------------------------------------
def describe(*numbers, decimals=2, label="Data"):
    """Return (count, min, max, mean) of any number of values and print them."""
    if len(numbers) == 0:
        print(f"{label}: NO values")
        return None

    print(f"{label} : n = len(numbers), min = {min(numbers)}, max = {max(numbers)}, mean = {mean(numbers)}")
    return len(numbers), min(numbers), max(numbers), mean(numbers)


def make_tag(text, **style):

    attrs =""
    for key , value in style.items():
        attrs += f' {key}="{value}"'
    return f"<span{attrs}>{text}</span>"

print()
describe(12, 7, 30, 4)
describe(3.14159, 2.71828, decimals=3, label="Constants")
describe()
print(make_tag("Hello", color="red", size=14))

square = lambda x: x ** 2
is_adult = lambda age: age >= 18
print(square(9), is_adult(17), is_adult(21))
```

```python
def electricity_bill(kwh):

    if kwh < 0:
        print(f"{kwh} is invalid. Please enter a non-negative value.")
        return None
    
    tier1 = min(kwh,50)
    tier2 = min(max(kwh-50, 0),150)
    tier3 = max(kwh-200, 0)

    bill = tier1 * 480 + tier2 * 610 + tier3 * 730

    "plus tax 10% and fee 2000"
    bill += 2000 
    bill += 0.1*bill
    return bill


for kwh in [0, 35, 50, 120, 200,350]:
    bill = electricity_bill(kwh)
    if bill is not None:

        print(f"{kwh} KWH -> {bill:,} riel (≈ {bill /4100:.2f} USD) ")


def main():
    print("============================================================")
    kwh = input("Enter electricity consumption in KWH: ").strip()
    try:
        kwh = float(kwh)
        bill = electricity_bill(kwh)
        if bill is not None:
            print(f"{kwh} KWH -> {bill:,} riel (≈ {bill /4100:.2f} USD) ")
    except ValueError:
        print("Invalid input. Please enter a valid number.")
if __name__ == "__main__":
    main()
```

```python
def is_prime(n):

    if n < 2:
        return False
    d=2
    while d * d <= n:
        
        if n%d ==0:
            return False
        d+=1

    return True


print("print all prime numbers below 50 :", end=" ")


for i in range(50):
    if is_prime(i) == True:
        print(i,end=" ")


print("\nNumber of primes below 100 :", end =" ")
count = 0

for i  in range(100):
    if is_prime(i) == True:
        count +=1

print(count)

def goldbach(n):

    for p in range(2 , n//2+1):

        if is_prime(p) and is_prime(n-p):

            return p,n-p
    return None


for n in range(4, 31, 2):
    p, q = goldbach(n)
    print(f"{n} = {q} + {p}")


all_ok =True
for n in range(4,10001,2):

    if goldbach (n) is None:
        all_ok =False
        print("counter,example:", n)
        break
print("Goldbach holds for every even number from 4 to 10000:", all_ok)
```
